## Exercise 1 — Café checkout

A campus café wants a tiny checkout calculation.

A customer buys:
- `3` coffees at £3.20 each;
- `2` sandwiches at £4.75 each.

If the **subtotal is at least £15**, the customer receives a **10% discount**.

### Your task

Using variables, arithmetic and an `if` statement:

1. Calculate the subtotal.
2. Calculate the final total after any discount.
3. Store either `"discount applied"` or `"no discount"` in a variable called `message`.
4. Print the subtotal, final total and message.

For the data above, the final total should be **£17.19**.

*Tip: `round(value, 2)` rounds a number to two decimal places.*


````python
coffee_price = 3.20
sandwich_price = 4.75
n_coffees = 3
n_sandwiches = 2

subtotal = n_coffees * coffee_price + n_sandwiches * sandwich_price

if subtotal >= 15:
    final_total = subtotal * 0.90
    message = "discount applied"
else:
    final_total = subtotal
    message = "no discount"

subtotal = round(subtotal, 2)
final_total = round(final_total, 2)

print("Subtotal:", subtotal)
print("Final total:", final_total)
print(message)
````

**Why this works:** the calculation is stored first, then the conditional chooses exactly one branch. The threshold is `>= 15`, so exactly £15 also gets the discount.

## Exercise 2 — Mobile-data warning

A phone provider wants a function that labels a customer's monthly data usage.

Complete:

```python
usage_status(used_gb, allowance_gb)
```

The function should return:

- `"over limit"` if `used_gb` is greater than `allowance_gb`;
- `"near limit"` if the customer has used **at least 80%** of the allowance but has not gone over;
- `"within limit"` otherwise.

### Examples

```python
usage_status(11, 10)   # "over limit"
usage_status(8, 10)    # "near limit"
usage_status(4, 10)    # "within limit"
```


````python
def usage_status(used_gb, allowance_gb):
    if used_gb > allowance_gb:
        return "over limit"
    elif used_gb >= 0.8 * allowance_gb:
        return "near limit"
    else:
        return "within limit"

assert usage_status(11, 10) == "over limit"
assert usage_status(8, 10) == "near limit"
assert usage_status(7.9, 10) == "within limit"
assert usage_status(10, 10) == "near limit"

print("All checks passed.")
````

**Key idea:** check the over-limit case first. Otherwise an over-limit customer would also satisfy the 80% condition.

## Exercise 3 — Saving for a laptop

A student has some money already saved and adds the same amount every month.

Write:

```python
months_to_target(starting_amount, monthly_saving, target)
```

The function should use a **`while` loop** and return the number of months required to reach or exceed the target.

### Examples

```python
months_to_target(200, 100, 500)  # 3
months_to_target(500, 100, 500)  # 0
months_to_target(450, 30, 500)   # 2
```

You may assume `monthly_saving` is positive.


````python
def months_to_target(starting_amount, monthly_saving, target):
    current_amount = starting_amount
    months = 0

    while current_amount < target:
        current_amount = current_amount + monthly_saving
        months = months + 1

    return months

assert months_to_target(200, 100, 500) == 3
assert months_to_target(500, 100, 500) == 0
assert months_to_target(450, 30, 500) == 2

print("All checks passed.")
````

**Key idea:** a `while` loop is useful because we do not know the required number of repetitions in advance.

## Exercise 4 — Delivery-time audit

A delivery company records delivery times in minutes:

```python
times = [31, 28, 45, 36, 25, 52, 33]
```

The target is 35 minutes.

### 4a — Find the late deliveries

Write:

```python
late_deliveries(times, target)
```

It should return a **new list** containing only delivery times that are greater than the target.

Example:

```python
late_deliveries([31, 28, 45, 36], 35)
# [45, 36]
```

### 4b — Find the slowest delivery

Write:

```python
slowest_index(times)
```

It should return the index of the **first occurrence** of the largest value.

Example:

```python
slowest_index([31, 52, 45, 52])
# 1
```

Try to solve both parts with loops rather than using `max()` or `.index()`.


````python
def late_deliveries(times, target):
    late = []
    for time in times:
        if time > target:
            late.append(time)
    return late


def slowest_index(times):
    largest_index = 0
    for i in range(1, len(times)):
        if times[i] > times[largest_index]:
            largest_index = i
    return largest_index


assert late_deliveries([31, 28, 45, 36], 35) == [45, 36]
assert late_deliveries([10, 20], 30) == []
assert slowest_index([31, 52, 45, 52]) == 1
assert slowest_index([9, 3, 2]) == 0

print("All checks passed.")
````

**Why the first maximum is returned:** the index changes only when a strictly larger value is found. An equal value later in the list does not replace the first one.

## Exercise 5 — Debugging a temperature monitor

The following function is supposed to return the highest temperature in a list.

It appears to work on some examples, but it contains a bug.

```python
def highest_temperature(readings):
    highest = 0
    for temperature in readings:
        if temperature > highest:
            highest = temperature
    return highest
```

### Your task

1. Run the function on the two examples below.
2. Explain why the second answer is wrong.
3. Fix the function.

```python
highest_temperature([12, 18, 15])      # should be 18
highest_temperature([-8, -3, -11])     # should be -3
```

Assume the list contains at least one value.


````python
def highest_temperature(readings):
    highest = readings[0]
    for temperature in readings:
        if temperature > highest:
            highest = temperature
    return highest

print(highest_temperature([12, 18, 15]))
print(highest_temperature([-8, -3, -11]))

assert highest_temperature([12, 18, 15]) == 18
assert highest_temperature([-8, -3, -11]) == -3
````

**Bug:** starting with `0` assumes that at least one reading is non-negative. Starting with the first actual observation works for negative values too.

## Exercise 6 — Searching customer order numbers

An online shop stores order numbers in a list.

### 6a — Linear search

Write:

```python
linear_order_search(orders, target)
```

Return the index of the **first** matching order number, or `-1` if the order is not present.

```python
linear_order_search([104, 108, 115, 121], 115)  # 2
linear_order_search([104, 108, 115, 121], 999)  # -1
```

### 6b — Binary search with a counter

If the order numbers are sorted, write:

```python
binary_order_search(orders, target)
```

Return a tuple:

```python
(found, iterations)
```

where `found` is `True` or `False`, and `iterations` counts how many times the `while` loop runs.

Examples:

```python
binary_order_search([101, 105, 110, 115, 120], 110)
# (True, 1)

binary_order_search([101, 105, 110, 115, 120], 999)
# (False, 3)
```


````python
def linear_order_search(orders, target):
    for i in range(len(orders)):
        if orders[i] == target:
            return i
    return -1


def binary_order_search(orders, target):
    low = 0
    high = len(orders) - 1
    iterations = 0

    while low <= high:
        iterations = iterations + 1
        midpoint = (low + high) // 2
        guess = orders[midpoint]

        if guess == target:
            return True, iterations
        elif guess > target:
            high = midpoint - 1
        else:
            low = midpoint + 1

    return False, iterations


assert linear_order_search([104, 108, 115, 121], 115) == 2
assert linear_order_search([104, 108, 115, 121], 999) == -1
assert binary_order_search([101, 105, 110, 115, 120], 110) == (True, 1)
assert binary_order_search([101, 105, 110, 115, 120], 999) == (False, 3)

print("All checks passed.")
````

**Key comparison:** linear search can be used on an unsorted list and may inspect every element. Binary search repeatedly halves the remaining search area, but requires sorted data.

## Exercise 7 — Which algorithm scales better?

For each code fragment, state the most appropriate Big-O category from:

- `O(1)`
- `O(n)`
- `O(n^2)`
- `O(log n)`

Assume `values` has length `n`.

### A

```python
first = values[0]
```

### B

```python
for value in values:
    print(value)
```

### C

```python
for x in values:
    for y in values:
        print(x, y)
```

### D

A binary search on a sorted list of length `n`.

Enter your answers in the strings below.


````python
complexity_A = "O(1)"
complexity_B = "O(n)"
complexity_C = "O(n^2)"
complexity_D = "O(log n)"

print("A:", complexity_A)
print("B:", complexity_B)
print("C:", complexity_C)
print("D:", complexity_D)
````

### Explanation

- **A — `O(1)`**: one list access, regardless of list length.
- **B — `O(n)`**: one pass through the list.
- **C — `O(n^2)`**: for each of the `n` outer-loop iterations, the inner loop also performs `n` iterations.
- **D — `O(log n)`**: binary search repeatedly halves the remaining search space.

Timing one run on one machine is not a general description of an algorithm: execution time depends on the computer, background processes, implementation details and the particular input. Big-O instead describes how the amount of work grows as the input size grows.

# Optional challenge — Detecting the first repeated code

A ticketing system receives a list of integer access codes. You want to find the **first value whose second occurrence appears earliest**.

Write:

```python
first_repeat(codes)
```

The function should return the repeated value. If nothing repeats, return `None`.

### Examples

```python
first_repeat([7, 4, 9, 4, 7])   # 4
first_repeat([2, 5, 2, 5])      # 2
first_repeat([1, 2, 3])         # None
```

### Rules

For this challenge, use only concepts covered so far: lists, loops, conditionals, indexing and functions.

Do **not** use sets or dictionaries.

### Think before coding

One possible strategy is:

- inspect each element in order;
- look back through the elements that appeared before it;
- return as soon as you find a match.

After you get it working, answer:

1. Why does this solution need nested loops?
2. What is its worst-case Big-O complexity?


````python
def first_repeat(codes):
    for i in range(len(codes)):
        for j in range(i):
            if codes[i] == codes[j]:
                return codes[i]
    return None


assert first_repeat([7, 4, 9, 4, 7]) == 4
assert first_repeat([2, 5, 2, 5]) == 2
assert first_repeat([1, 2, 3]) is None

print("All checks passed.")
````

### Challenge explanation

When we reach position `i`, the inner loop checks whether the current value appeared at any earlier position `0, 1, ..., i-1`. Returning immediately ensures we return the value whose **second occurrence appears first**.

In the worst case, the algorithm compares many pairs of elements. The number of comparisons grows roughly in proportion to `n^2`, so the worst-case complexity is **`O(n^2)`**.

Later in the course, additional data structures can make duplicate checking much more efficient.
